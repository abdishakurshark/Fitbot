Understanding NLP Techniques
Before implementation:

Tokenization: Breaking text into individual words or sentences

Stopword Removal: Filtering out common words that don't add meaning (e.g., "the", "and")

Tagging (POS Tagging): Identifying parts of speech for each word

Lemmatization: Reducing words to their base/dictionary form (e.g., "running" → "run")

Stemming: Crudely chopping off word endings (e.g., "running" → "run")

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag

# Download required NLTK data
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')

def process_text(text):
    print("Original Text:\n", text)

    # Tokenization
    words = word_tokenize(text)
    sentences = sent_tokenize(text)
    print("\nTokenized Words:", words)
    print("\nTokenized Sentences:", sentences)

    # Stopword Removal
    stop_words = set(stopwords.words('english'))
    filtered_words = [word for word in words if word.lower() not in stop_words and word.isalpha()]
    print("\nAfter Stopword Removal:", filtered_words)

    # POS Tagging
    tagged_words = pos_tag(filtered_words)
    print("\nPOS Tags:", tagged_words)

    # Stemming
    stemmer = PorterStemmer()
    stemmed_words = [stemmer.stem(word) for word in filtered_words]
    print("\nStemmed Words:", stemmed_words)

    # Lemmatization
    lemmatizer = WordNetLemmatizer()
    lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
    print("\nLemmatized Words:", lemmatized_words)

    return {
        'words': words,
        'sentences': sentences,
        'filtered_words': filtered_words,
        'tagged_words': tagged_words,
        'stemmed_words': stemmed_words,
        'lemmatized_words': lemmatized_words
    }

# Fitness-related text for FitBot
fitness_text = """
FitBot can help you achieve your fitness goals.
Running 3 times a week and doing strength training exercises
like push-ups and squats will improve your endurance and muscle strength.
Remember to stay hydrated and eat balanced meals for optimal results.
"""

# Process the text
processed_data = process_text(fitness_text)
```

# Phase 2: Building Classifiers for FitBot

In this phase, we aim to implement two classification models — Decision Tree and Naive Bayes — to categorize user queries related to fitness and nutrition. We will also evaluate their performance and explore ways to improve their accuracy.


## Step 1: Understanding Decision Tree and Naive Bayes Classifiers

- **Decision Tree Classifier**: A model that splits data into branches based on feature values. It is easy to interpret but may overfit small datasets.
- **Naive Bayes Classifier**: A probabilistic model that assumes feature independence. It is simple, fast, and works well for text classification tasks.


## Step 2: Preparing the Dataset

We create a small sample dataset with labeled user queries related to fitness goals. Each sentence belongs to a specific category such as weight loss, workout, nutrition, or progress tracking.


```python
# Sample data and labels
sample_data = [
    "I want to lose weight.",
    "Suggest a good workout for beginners.",
    "What should I eat to gain muscle?",
    "Can you help me track my fitness progress?",
    "How often should I train each week?"
]

sample_labels = [
    "weight_loss",
    "workout",
    "nutrition",
    "progress_tracking",
    "workout"
]
```

## Step 3: Feature Extraction Using TF-IDF

We convert the text data into numerical features using the TF-IDF Vectorizer. TF-IDF (Term Frequency-Inverse Document Frequency) represents the importance of words relative to the document and the entire dataset.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(sample_data)
y = sample_labels
```

## Step 4: Training Decision Tree and Naive Bayes Models

We train two models:
- Decision Tree Classifier
- Multinomial Naive Bayes Classifier


```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import MultinomialNB

# Initialize the models
decision_tree_model = DecisionTreeClassifier()
naive_bayes_model = MultinomialNB()

# Train the models
decision_tree_model.fit(X, y)
naive_bayes_model.fit(X, y)
```

## Step 5: Evaluating the Models

We predict on the training data and evaluate the models based on accuracy and confusion matrices.


```python
from sklearn.metrics import accuracy_score, confusion_matrix
import pandas as pd

# Predictions
dt_predictions = decision_tree_model.predict(X)
nb_predictions = naive_bayes_model.predict(X)

# Evaluation
print("🔵 Decision Tree Accuracy:", accuracy_score(y, dt_predictions))
print("🔵 Decision Tree Confusion Matrix:\n", pd.DataFrame(confusion_matrix(y, dt_predictions)))

print("\n🟢 Naive Bayes Accuracy:", accuracy_score(y, nb_predictions))
print("🟢 Naive Bayes Confusion Matrix:\n", pd.DataFrame(confusion_matrix(y, nb_predictions)))
```

## Step 6: Ways to Improve the Accuracy

To improve the model's performance on real-world datasets:
- Increase the size of the dataset
- Use train/test split or cross-validation
- Tune hyperparameters such as `max_depth` for Decision Tree and `alpha` for Naive Bayes
- Perform better text preprocessing (e.g., removing punctuation, using bigrams)


### PHASE 3
### 📖 What is Hyperparameter Tuning?
Hyperparameters are external configurations set before training a machine learning model.  
Proper tuning helps a model generalize better and improves its accuracy.

- 🔵 **Decision Tree Important Hyperparameters:**
  - `criterion` (splitting quality: 'gini' or 'entropy')
  - `max_depth` (maximum depth of the tree)
  - `min_samples_split` (minimum samples required to split an internal node)

- 🟢 **Naive Bayes Important Hyperparameter:**
  - `var_smoothing` (smoothing parameter for numeric stability)

In this phase, we will perform hyperparameter tuning and evaluate its effect.


### 🛠️ Step 2: Implementing Hyperparameter Tuning

We will:
- Train initial models for Decision Tree and Naive Bayes without tuning
- Apply hyperparameter tuning using GridSearchCV
- Retrain models with the best-found parameters
- Compare the performance before and after tuning


### 🌳 Step 3: Training and Evaluating Initial Decision Tree

We first train a Decision Tree without any hyperparameter tuning and evaluate its accuracy and confusion matrix.


### 🛠️ Step 4: Tuning the Decision Tree

We perform Grid Search with cross-validation to find the best parameters for Decision Tree.
Parameters to tune:
- `criterion`: 'gini' or 'entropy'
- `max_depth`: various depths
- `min_samples_split`: minimum number of samples required to split a node


### 🧠 Step 5: Training and Evaluating Initial Naive Bayes

We train a Gaussian Naive Bayes model without tuning and evaluate it.


### 🛠️ Step 6: Tuning the Naive Bayes Model

For Naive Bayes, we tune the `var_smoothing` parameter, which controls numerical stability.


```python

import pandas as pd
from sklearn.model_selection import GridSearchCV

# Hyperparameter tuning for Decision Tree
param_grid_dt = {
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

grid_search_dt = GridSearchCV(decision_tree_model, param_grid_dt, cv=2, scoring='accuracy')
grid_search_dt.fit(X, y)

print("Best hyperparameters for Decision Tree:", grid_search_dt.best_params_)
print("Best accuracy for Decision Tree:", grid_search_dt.best_score_)

# Hyperparameter tuning for Naive Bayes
param_grid_nb = {
    'alpha': [0.1, 0.5, 1.0, 2.0]
}

grid_search_nb = GridSearchCV(naive_bayes_model, param_grid_nb, cv=2, scoring='accuracy')
grid_search_nb.fit(X, y)

print("\nBest hyperparameters for Naive Bayes:", grid_search_nb.best_params_)
print("Best accuracy for Naive Bayes:", grid_search_nb.best_score_)

# Evaluate tuned models
tuned_dt_predictions = grid_search_dt.predict(X)
tuned_nb_predictions = grid_search_nb.predict(X)

print("\nTuned Decision Tree Accuracy:", accuracy_score(y, tuned_dt_predictions))
print("Tuned Decision Tree Confusion Matrix:\n", pd.DataFrame(confusion_matrix(y, tuned_dt_predictions)))

print("\nTuned Naive Bayes Accuracy:", accuracy_score(y, tuned_nb_predictions))
print("Tuned Naive Bayes Confusion Matrix:\n", pd.DataFrame(confusion_matrix(y, tuned_nb_predictions)))

# Analyze and compare performance
print("\nPerformance Comparison:")
print("Decision Tree - Initial Accuracy:", accuracy_score(y, dt_predictions), "Tuned Accuracy:", accuracy_score(y, tuned_dt_predictions))
print("Naive Bayes - Initial Accuracy:", accuracy_score(y, nb_predictions), "Tuned Accuracy:", accuracy_score(y, tuned_nb_predictions))
```

### 📊 Step 7: Final Results and Analysis

#### Decision Tree:
- Initial Accuracy: (1.0)
- Tuned Accuracy: (1.0)
- Improvement: No change after tuning. Model maintained perfect accuracy, possibly due to overfitting on the dataset.

#### Naive Bayes:
- Initial Accuracy: (1.0)
- Tuned Accuracy: (1.0)
- Improvement: No change after tuning. Model achieved perfect accuracy after tuning, but this may also suggest overfitting.

✅ Conclusion: Both Decision Tree and Naive Bayes achieved perfect accuracy after tuning. However, the sudden jump from poor initial performance (especially for Naive Bayes) to perfect results indicates potential overfitting, suggesting that model validation on unseen data is necessary.
